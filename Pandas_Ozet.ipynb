```python
# ! 1. Temel Python (Veri Yapıları ve Algoritma)

# 1. Değişkenler ve Metin Biçimlendirme (f-string)
isim = "Nacican"
yas = 23
print(f"Benim adım {isim}, yaşım {yas}.")

# 2. Listeler ve Liste Metotları
fiyatlar = [100, 105, 102]
fiyatlar.append(110) # Listenin sonuna yeni eleman ekler
en_yuksek = max(fiyatlar) # Listedeki en büyük değeri bulur

# 3. Sözlükler (Key-Value Mantığı)
varlik = {"Sembol": "BTC", "Fiyat": 62000}
varlik["Fiyat"] = 62500 # Var olan bir değeri günceller

# 4. Koşullar (If / Elif / Else)
getiri = 5.5
if getiri > 0:
    print("Kârdayız")
elif getiri < 0:
    print("Zarardayız")
else:
    print("Başa Baş")

# 5. Döngüler (For Loop)
hedef_fiyatlar = [] # Boş liste
for fiyat in fiyatlar:
    if fiyat > 102:
        hedef_fiyatlar.append(fiyat) # Şartı sağlayanları yeni listeye atar
```

```python
# 2. Pandas Temelleri ve Sütun İşlemleri

import pandas as pd

# 1. Sözlükten DataFrame (Tablo) Oluşturma
veri = {
    "Coin": ["BTC", "ETH", "SOL"],
    "Fiyat": [61000, 3100, 140],
    "Adet": [2, 10, 50]
}
df = pd.DataFrame(veri)

# 2. Matematiksel İşlem ile Yeni Sütun Ekleme
# İki sütunu birbiriyle çarpıp yeni bir sütuna eşitliyoruz
df["Toplam_Yatirim"] = df["Fiyat"] * df["Adet"]

# 3. Formül Uygulama ve Yuvarlama (.round)
df["Hedef_Fiyat"] = (df["Fiyat"] * 1.05).round(2)
```

```python
# 3. Filtreleme ve Sıralama (Screener Mantığı)

# 1. Çoklu Şartla Filtreleme (Ve: & / Veya: |)
# Kural: Her şart mutlaka kendi normal parantezi () içinde olmalı!
filtrelenmis_tablo = df[(df["Toplam_Yatirim"] > 50000) & (df["Fiyat"] > 1000)]

# 2. Tabloyu Sıralama (.sort_values)
# ascending=False (Büyükten küçüğe), ascending=True (Küçükten büyüğe)
sirali_tablo = filtrelenmis_tablo.sort_values(by="Toplam_Yatirim", ascending=False)
```

```python
# 4. Dış Dünya ile İletişim (Dosya Okuma / Yazma)

# 1. Bilgisayardaki CSV dosyasını Pandas tablosuna dönüştürme
df_disari = pd.read_csv("piyasa.csv")

# 2. İşlenmiş Pandas tablosunu bilgisayara yeni bir dosya olarak kaydetme
# index=False komutu, baştaki anlamsız 0,1,2 sıra numaralarının kaydedilmesini önler
sirali_tablo.to_csv("hazir_rapor.csv", index=False)
```

```python
# 5. Zaman Serileri ve Dilimleme (.loc Mantığı)

# 1. Tarih sütununu normal metinden gerçek zaman formatına çevirme
df["Tarih"] = pd.to_datetime(df["Tarih"])

# 2. Tarihi tablonun indeksine (satır başlarına) sabitleme
df = df.set_index("Tarih")

# 3. Belirli bir günün verisini çekme (.loc)
hedef_gun = df.loc["2026-09-25"]

# 4. İki tarih arasındaki dönemi kesip alma (Slicing / Dilimleme)
hedef_aralik = df.loc["2026-09-24":"2026-09-26"]
```
