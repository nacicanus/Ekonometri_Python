```python
# 1. Dosyayı her zamanki gibi okutuyoruz
df_gecmis = pd.read_csv("gecmis_veri.csv")

# 2. Tarih sütununu normal metinden, Pandas'ın anladığı 'gerçek zaman' formatına çeviriyoruz
df_gecmis["Tarih"] = pd.to_datetime(df_gecmis["Tarih"])

df_gecmis = df_gecmis.set_index("Tarih")
print(df_gecmis)

```

```python
# Pandas'ta indeks üzerinden belirli bir satırı doğrudan çekip almak için İngilizce "locate" (konum bul) kelimesinin kısaltması olan .loc[] (lokasyon) aracını kullanırız.
# secilen_veri = tablo_adi.loc["Aranan_İndeks_Adı"]

hedef_gun = df_gecmis.loc["2026-09-25"]

print(hedef_gun)
```

```python
# Python'da "başlangıç noktasından bitiş noktasına kadar olan her şeyi al"
# demek için araya iki nokta üst üste (:) koyarız. Bu kavrama dilimleme (slicing) denir.
# Eğer tarihi indeks yapmasaydık, iki tarih arasını çekmek için çok karmaşık küktür/büyüktür koşulları yazmamız gerekirdi.
# Ama artık indeksimiz tarih olduğuna göre mantık çok basit:
# secilen_aralik = tablo_adi.loc["baslangic_tarihi" : "bitis_tarihi"]

hedef_aralik = df_gecmis.loc["2026-09-24":"2026-09-26"]
print(hedef_aralik)
```

```python
# Pandas'ta geçmiş verileri bir adım (veya istediğin kadar adım) aşağı kaydırmak için .shift() metodunu kullanırız.

print("Merhaba github")
print("Merhaba github 2")
print("Merhaba github 3")
```

```python

df_gecmis["Onceki_Kapanis"] = df_gecmis["Kapanis"].shift(1)
df_gecmis["Günlük_Getiri"] = (((df_gecmis["Kapanis"] - df_gecmis["Onceki_Kapanis"]) / df_gecmis["Onceki_Kapanis"]) * 100).round(2)


print(df_gecmis)
```

```python
# Mantık Dersi 6: Geçmişe Doğru Pencere Açmak (.rolling)

# "Son 3 günün ortalaması nedir?" diye sormak için de Pandas'ın pencere (rolling) mantığını kullanırız.
# Python'da belirli bir dönemin ortalamasını almak (SMA - Simple Moving Average) için şu kalıp kullanılır:
# tablo["Yeni_Sütun"] = tablo["Hedef_Sütun"].rolling(window=Gün_Sayısı).mean()

# Buradaki .rolling(window=3) kısmı "bana son 3 satırlık bir pencere aç" derken, 
# sonundaki .mean() ise "bu penceredeki değerlerin ortalamasını al" (matematikteki mean/ortalama) anlamına gelir.

df_gecmis["SMA_3"] = df_gecmis["Kapanis"].rolling(window=3).mean().round(2)
print(df_gecmis)
```

```python
# Mantık Dersi 7: Karar Algoritmaları (Koşullu Sütunlar)

# Excel'deki EĞER (IF) formülünü düşün. Algoritmaya şunu demeliyiz: 
# "Eğer bugünkü fiyat ortalamanın üzerindeyse trend yükseliştedir, 'AL' yaz. Altındaysa 'BEKLE' yaz."
# Pandas'ta bu tarz koşullu durumları (eğer böyleyse bunu yaz, değilse şunu yaz) en hızlı şekilde çözmek için
# Python'un matematik ve veri işlemlerindeki diğer süper gücü olan NumPy kütüphanesini kullanırız.
# NumPy'ın where (nerede/durumunda) adlı aracı tam bir karar motorudur.

# tablo["Yeni_Sutun"] = np.where(Koşul, "Şart Sağlanırsa Yazılacak Şey", "Sağlanmazsa Yazılacak Şey")
# Örneğin: np.where(df["Fiyat"] > 100, "Pahalı", "Ucuz")

import numpy as np

df_gecmis["Sinyal"] = np.where(df_gecmis["Kapanis"] > df_gecmis["SMA_3"], "AL", "BEKLE")
print(df_gecmis)
```

```python
# Mantık Dersi 8: Veri Temizleme (.dropna)
# Bunun komutu İngilizce "düşür/at" anlamına gelen .dropna() metodudur.

# Mantık Dersi 9: Sayma ve Özetleme (.value_counts)
# Tabloyu temizledikten sonra işin analitik kısmı başlar. Örneğin bir strateji kurdun ve şu soruyu sorman gerekiyor: 
# "Peki bu algoritma test ettiğim tarih aralığında toplam kaç gün AL, kaç gün BEKLE demiş?"
# Bir sütunun içindeki benzersiz verilerin 
# (AL ve BEKLE yazılarının) kaçar kez tekrar ettiğini tek kalemde saymak için .value_counts() komutunu kullanırız.

df_gecmis = df_gecmis.dropna()

sinyal_sayisi = df_gecmis["Sinyal"].value_counts()

print(sinyal_sayisi)

```
